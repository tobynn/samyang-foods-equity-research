# 01. Historical Financial Trend Analysis

This notebook is the first executable step of the Samyang Foods project.

**Goal**
- load the official 2021-2025 headline financial dataset,
- calculate growth and margins,
- identify questions that require deeper filing-level research.


```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

pd.options.display.float_format = "{:,.2f}".format

DATA_PATH = Path("../data/processed/financials_annual.csv")
df = pd.read_csv(DATA_PATH)
df
```

```python
df["revenue_growth"] = df["revenue_krw_mn"].pct_change()
df["op_margin"] = df["operating_profit_krw_mn"] / df["revenue_krw_mn"]
df["net_margin"] = df["net_income_krw_mn"] / df["revenue_krw_mn"]

years = int(df["year"].iloc[-1] - df["year"].iloc[0])
revenue_cagr = (df["revenue_krw_mn"].iloc[-1] / df["revenue_krw_mn"].iloc[0]) ** (1 / years) - 1

summary = df[[
    "year", "revenue_krw_mn", "operating_profit_krw_mn",
    "net_income_krw_mn", "revenue_growth", "op_margin", "net_margin"
]].copy()

summary
```

```python
print(f"2021-2025 Revenue CAGR: {revenue_cagr:.1%}")
print(f"2025 Operating Margin: {df['op_margin'].iloc[-1]:.1%}")
print(f"2025 Net Margin: {df['net_margin'].iloc[-1]:.1%}")
```

```python
fig, ax = plt.subplots(figsize=(8, 4.5))
ax.plot(df["year"], df["revenue_krw_mn"], marker="o")
ax.set_title("Samyang Foods Revenue")
ax.set_xlabel("Year")
ax.set_ylabel("KRW mn")
ax.ticklabel_format(style="plain", axis="y")
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(8, 4.5))
ax.plot(df["year"], df["op_margin"] * 100, marker="o", label="Operating margin")
ax.plot(df["year"], df["net_margin"] * 100, marker="o", label="Net margin")
ax.set_title("Profitability Trend")
ax.set_xlabel("Year")
ax.set_ylabel("%")
ax.legend()
plt.show()
```
