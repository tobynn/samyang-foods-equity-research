```python
fig, ax = plt.subplots(figsize=(8, 4.5))
p = drivers[drivers['period'].isin(['2025A','2026Q1','2026Q2'])]
ax.plot(p['period'], p['overseas_share']*100, marker='o')
ax.set_title('Overseas Sales Share')
ax.set_ylabel('% of consolidated revenue')
ax.set_ylim(75, 86)
plt.show()
```
